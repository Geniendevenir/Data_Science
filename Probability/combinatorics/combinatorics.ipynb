## VOCABULARY

### Ordered vs Unordered
**Order matters:** The position matters. If you pick numbers for a password like 1, 2, 3, that is completely different from typing 3, 2, 1. The slot they occupy (first, second, third) defines the outcome. 

**Order doesnt matters:** Only the final "bucket" or collection matters. If you are choosing a group of 3 books from a stack, grabbing Book A then Book B then Book C results in the exact same pile of books as grabbing Book C then Book A then Book B. In an unordered world, (A, B, C) and (C, A, B) are the exact same thing.

Example: Say we pick teamates for a sport team and want to pick (Alice and Hugo).\
	-> Order matters if the other teams can pick after our first choice \
	-> Order doesnt matter if: we pick every teamates in a row.


### Distinct vs Indistinct
**Distinct (Unique):** All the objects of the set are unique and can be differentiated. \
Note: In probabilites we can "make" indistinct objects, distinct if we want.

**Partially Indistinct:** Means that accross all objects in a set, some of them are the sames. Like the characters in the word "MISSISSIPI".

**Indistinct:** All objects of the set looks the same and can not be differentiated. \
Think of a string of bits. They can all be 0 or 1.

Example: There are 5 toys in a box. 3 Cows and 2 pigs. We need to pick 4 toys.
	-> When can choose to make the toys Distinct as if there was a difference between Cow 1 and Cow 2.
	-> Or let them as (partially) indistinct: the cows looks indentical and the pigs too.


### Replaced vs Not Replaced
**Replaced:** If we put back the objects in the set after picking it. It can thus be picked again.

**Not Replaced:** If we don't put back the objects once it has been picked.

Example: Lets say we have 5 unique toys and we need to pick 2:
	-> If we replace the toys, we can pick the exact same toy twice
	-> Otherwise we are forced to pick 2 different toys.

## Permutation: Distinct / Ordered

### Distinct Permutation
**Definition:** Ordered arrangment of n distinct object.


**Formula:** $$n!$$

**Examples** \
1 - How many distinct ordered arrangment of characters are there in the string "BAYES" ? $$5!$$
2 - A smartphon has a 4 digits password. Suppose there are 4 smudges over 4 digits. How many distinct passcodes are possible ? $$4!$$

```python
# 1 - Get all 5! permutations

bayes_all = list(itertools.permutations(["B", "A", "Y", "E", "S"]))
bayes_unique = list(set(itertools.permutations(["B", "A", "Y", "E", "S"])))

# Check that permuations returns unique combinations
print(len(bayes_all) == len(bayes_unique))

# 2 - Get all 4! permutations:
passcode_4 = list(itertools.permutations([1, 2, 3, 4]))
print(len(passcode_4))
```

## Permutation: Partially Indistinct / Ordered

### Partially Indistinct Permutation
**Definition:** Ordered arrangment of n object some of them being indistincts.


**Formula:** $$\frac{n!}{n_0!n_1!...n_r!}$$

**Examples** \
3 - How many distinct ordered arrangment of characters are there in the string "MISSISSIPI" ? $$\frac{10!}{4!4!}$$
4 - A smartphon has a 4 digits password. Suppose there are 3 smudges over 4 digits. How many distinct passcodes are possible ? $$\frac{4!}{2!}$$

```python
# 3 - "MISSISSIPI" Number of permutations
numerator = math.factorial(10)
denominator = math.factorial(4) * math.factorial(4)

result = numerator / denominator
print(result)

# 4 - 3 Smudges / 4 digit passcode
numerator = math.factorial(4)
denominator = math.factorial(2)

result = numerator / denominator
print(result)
```

## Combination: Distinct / Unordered

### 
**Definition:** An unordered selection of k objects from a set of n objects. \
The following formula is for the cases where the objects are distincts and are not replaced between experiment.


**Formula:** $${n \choose k} = \frac{n!}{k!(n - k)!}$$

**Examples** \
5 - In the Hunger Games, how many ways are there of choosing 2 villagers from district 12, which has a population of 8,000?
$${8000 \choose 2}$$
6 - How many ways are there to select 3 books out of 6, if there are two books that should not both be chosen together? \
For example, if you are choosing 3 out of 6 probability books, don't choose both the 8th and 9th edition of the Ross textbook.
$${2 \cdot {4 \choose 2} + {4 \choose 3}}$$

```python
# 5 - Hunger Games combinations
hunger = math.comb(8000, 2)
print(hunger)
```

## Bucketing: Distinct

### 
**Definition:** Placing n distinct objects into r containers. \
Note: a way to explain the followin formulas is we ave n steps and at each step we have r choices.

**Formula:** $$r^{n}$$

**Examples** \
7 - Say we are going to put 10 different strings into 5 buckets of a hash table. How many possible ways are there of doing this?
$$5^{10}$$

## Bucketing: Indistinct

### 
**Definition:** Placing n Indistinct objects into r containers.

**Formula:** $$\frac{(n + r - 1)!}{n!(r - 1)!} = {{n + r - 1}\choose{n}}$$

**Examples** \
8 - Say you are a startup incubator and you have $10 million to invest in 4 companies (in $1 million increments).\
How many ways can you allocate this money?
$$\frac{(10 + 4 - 1)!}{10!(4 - 1)!}$$
